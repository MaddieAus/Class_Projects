A. Write a one-line function ```is_palindromeV1(word)``` that determines whether word is a
palindrome, that is, reads the same forward and backward. For example,
is_palindromeV1('racecar') should return True, and ```is_palindromeV1('toyota')```
should return False. Assume that word is in all lowercase letters. Do not use any loops or function
calls, just slices.<br><br>
Hint:<br>
return x is y # or return x == y

B. Write a multi-line function ```is_palindromeV2(word)``` that determines whether word is a
palindrome using a loop. Do not use any function calls other than ```len()```, just slices in conjunction
with your loop.
Demonstrate that your function works with several inputs that pass and several that fail the test.<br><br>
C. Use the iPython ```%timeit``` magic command to measure and compare runtimes for
```is_palindromeV1()``` and ```is_palindromeV2()```. Run and compare using at least one input
that passes the test and at least one that fails the test. Describe the results.<br>

### <u>Answer:</u>

````python
# part A.
def is_palindromeV1(word):
    return word == word[::-1]  # backwords string and then compare
print(is_palindromeV1('racecar'))  # True
print(is_palindromeV1('madam'))    # True
print(is_palindromeV1('toyota'))   # False
print(is_palindromeV1('hello'))    # False
````

````python
# part B.
def is_palindromeV2(word):
    n = len(word)
    for i in range(n // 2):  # only check halfway
        if word[i] != word[n - i - 1]:  
            return False
    return True
print(is_palindromeV2('racecar'))  # True
print(is_palindromeV2('madam'))    # True
print(is_palindromeV2('toyota'))   # False
print(is_palindromeV2('hello'))    # False
````

### <u>Question 5.3.1</u>
In Sweden, dates are usually written in the “yyyy-mm-dd” format. For instance, 2020-09-01 for
September 1, 2020. Write and test a function ```ToSwedishDate(strDate)``` that converts a date from<br>
'm/d/yyyy' to 'yyyy-mm-dd'. NOTE: 'm/d/yyyy' includes the possibility of double-digit month and
date values.<br>
### <u>Answer:</u>

````python
def ToSwedishDate(strDate):
    month, day, year = strDate.split('/')
    return f"{year}-{int(month):02d}-{int(day):02d}"  #2 digits
#---test---
print(ToSwedishDate('9/1/2020'))   # should print 2020/09/01
print(ToSwedishDate('12/25/2025')) # should be 2025/12/25
````

### <u>Question 5.3.4</u>
A. Write a function ```get_digits(s)``` that returns a string of all the digits found in s (in the same order).
For example, ```get_digits(' **1.23a-42 ')``` should return '12342', and
```get_digits('LOL')``` should return '', the empty string.<br> <br>
B. Test your function with a variety of inputs to demonstrate that it works.<br><br>
C. Use the iPython %timeit magic command to measure and compare runtimes for
```get_digits()``` with different inputs. Describe the results.<br>
### <u>Answer:</u>

````python
# part A.
def get_digits(s):
    result = ''
    for char in s:
        if char.isdigit():
            result += char
    return result
# part B.
print(get_digits(' **1.23a-42 '))  # 12342
print(get_digits('LOL'))           # should be nothing
print(get_digits('2025 is cool!')) # 2025
````

````python
# part C.
%timeit get_digits(' **1.23a-42 ')
%timeit get_digits('LOL')
````

### <u>Question 5.4.5</u>
Write and test a function ```all_pairs(n)``` that returns the list of all ```𝑛(𝑛−1)```<br>
2 tuples (i, j), where 1 ≤ i<br>
< j ≤ n. For example, ```all_pairs(3)``` should return [ (1,2), (1,3), (2,3)].<br>
Demonstrate that your function works for several representative, reasonable values of n.<br>
### <u>Answer:</u>

````python
def all_pairs(n):
    pairs = []
    for i in range(1, n):
        for j in range(i + 1, n + 1):
            pairs.append((i, j))
    return pairs
#---test---
print(all_pairs(3))  # 3 pairs (1, 2), (1, 3), (2, 3)
print(all_pairs(4))  # 4 pairs (1, 2), (1, 3), (1, 4), (2, 3), (2, 4), (3, 4)
````

### <u>Question 5.4.6</u>
Pretend that list’s sort method does not exist and write your own function ```myselectionsort(a)```
that sorts list a in place in ascending order. Since the list is sorted in place, it should return None.<br>
Use the following algorithm, which is Selection Sort:<br>
```Select Sort(a):```
```Set n equal to len(a)```
Repeat the following steps ```while n > 1:```
Find the largest among the first n elements swap it with the nth element: a[n-1]<br>
Subtract 1 from n demonstrate that your ```myselectionsort(a)``` works with several representative, reasonable lists.<br>
### <u>Answer:</u>

````python
def myselectionsort(a):
    n = len(a)
    while n > 1:
        max_index = 0
        for i in range(1, n):
            if a[i] > a[max_index]:
                max_index = i
        a[max_index], a[n - 1] = a[n - 1], a[max_index]  # swap
        n -= 1
# ---test---
nums = [5, 2, 9, 1, 5, 6]
myselectionsort(nums)
print(nums)  #1, 2, 5, 5, 6, 9
````

### <u>Question 5.5.3</u>
Suppose in a dictionary all values associated with keys are different.<br>
Write a function
```reverse_dictionary(d)``` that takes such a dictionary and returns a “reverse” dictionary in which
each value becomes a key and its key becomes the associated value. For example,
```reverse_dictionary({1:'A', 2: 'B'})``` should return ```{'A':1, 'B':2}```.<br>
Do this in an iterative fashion, element by element.<br>
Demonstrate that your ```reverse_dictionary()``` function works with several representative,
reasonable dictionaries.<br>
### <u>Answer:</u>

````python
def reverse_dictionary(d):
    rev = {}
    for key, value in d.items():
        rev[value] = key
    return rev
#---test---
print(reverse_dictionary({1:'A', 2:'B'}))       # 'A':1, 'B':2
print(reverse_dictionary({'x':10, 'y':20}))     # 10:'x', 20:'y'
````

### <u>Question 5.6.5</u>
Write a function CountLetters(f) that counts the occurrences of the 26 English letters, case-blind,
in a text file f and prints their relative frequencies. The file is assumed to be in the same directory as your
Jupyter notebook.<br>
The function should take the file name f as an argument, count the number of
occurrences of each letter in the file, and display each letter, along with its count. The letters should be
displayed in order of frequency. For example, if the file contains two 'A' s, three 'B' s, and no other letters,
then the output should be:<br>
Text file name: letters.txt<br>
B: 3<br>
A: 2<br>
C: 0<br>
...<br>
Z: 0<br>
Hints:

1. Create a dictionary with 26 key-value pairs where the keys are 'A', 'B', 'C', ..., 'Z',
and the values are the counts, which are initialized to 0 (zero).

2. Read the file one line at a time. For each line after it is read, increment the appropriate counts
in the dictionary for each letter in the line.


3. Create a list of (count, letter) pairs. The built-in function zip(s1, s2) combines the
respective elements from the sequences s1 and s2 into pairs (tuples) and generates a sequence
of such tuples.


4. Sort the pairs: For example, sorted(zip([2, 3], ['A', 'B']), reverse=True)
produces [(3, 'B'), (2, 'A')].


5. Print out both elements of each pair in the desired format.
Download an ASCII text copy of the Wizard of Oz from here:<br>
https://github.com/kwartler/text_mining/blob/master/Wizard_Of_Oz.txt<br>
Save this file as Wizard_Of_Oz.txt in the same directory as your assignment Jupyter Notebook. (Do
this manually – do not write code in your Jupyter Notebook to download and save the file.) Downloading
via the “Raw” link on the GitHub page will download the text file.
Run your function on the Wizard_Of_Oz.txt file and show the output.<br>

### <u>Answer:</u>

````python
def CountLetters(f):
    # Step 1
    counts = {chr(c): 0 for c in range(ord('A'), ord('Z') + 1)}
    
    # Step 2
    with open(f, 'r', encoding='utf-8') as file:
        for line in file:
            for char in line.upper():  # changing to uppercase
                if char in counts:     # Only from A-Z
                    counts[char] += 1

    # Step 3
    freq_pairs = sorted(
        zip(counts.values(), counts.keys()), 
        reverse=True
    )

    # Step 4
    print(f"Wizard_Of_Oz: {f}")
    for count, letter in freq_pairs:
        print(f"{letter}: {count}")

#---test---
CountLetters('Wizard_Of_Oz.txt')
````
