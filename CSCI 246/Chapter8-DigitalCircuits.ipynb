### Question 8.3.8
Fill in the blanks in the function below, which returns a string of the eight hex digits (padded with zeros
on the left, if necessary) of any int value n such that 0 ≤ 𝑛 ≤ 232.
def my_hex(n):
"""
Return the string of eight hex digits of n.
"""
hex_digits = '0123456789abcdef'
s = ''
for k in range (___):
s = hex_digits [n & ___] + s
n >>= ___
return s

### Answer:

````python
def my_hex(n):
    """
    Return the string of eight hex digits of n.
    """
    hex_digits = '0123456789abcdef'
    s = ''
    for k in range(8):
        s = hex_digits[n & 0xf] + s
        n >>= 4
    return s
````

### Question 8.3.13
Bits 0-23 in a positive integer represent a line of pixels (picture elements) in a black and white image (the
remaining bits are 0s). 1 stands for black and 0 stands for white. We want to fill all single-pixel holes in
the line, so that each pixel with black left and right neighbors becomes black. For example,
011110111000011011010000
should become
011111111000011111110000
Fill in the blank in the following function that does this and returns an int corresponding to the new line
of pixels (with bits 0-23 updated as necessary and other bits set to 0).
def fill_holes(pix):
return _______

### Answer:

````python
def fill_holes(pix):
    mask = (1 << 24) - 1
    return (pix | ((pix << 1) & (pix >> 1))) & mask
````
