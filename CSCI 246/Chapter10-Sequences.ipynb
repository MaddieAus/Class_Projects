## Question 10.5.5
Fill the blanks in the code

### Answer

```python
def fibonacci_listV1(n):
    f = (n+1) * [0]
    f[1] = f[2] = 1
    for k in range(3, n+1):
        f[k] = f[k-1] + f[k-2]
    return f
```

```python

def fibonacci_listV2(n):
    f = [0, 1, 1]
    while len(f) <= n:
        f.append(f[-1] + f[-2])
    return f
```

```python

def fibonacci(n):
    if n == 0:
        return 0
    f1, f2 = 1, 1
    for _ in range(3, n+1):
        f1, f2 = f2, f1 + f2
    return f2
```
