### Question 6.2.4
Write 1510 and 2410 in base 3. Use Python base conversion functionality to do these same base conversions and check your answers.
<br>
### Answer:
1510 <br>
15 ÷ 3 = 5 remainder 0  <br>
5 ÷ 3 = 1 remainder 2  <br>

2410 <br>
24 ÷ 3 = 8 remainder 0  <br>
8 ÷ 3 = 2 remainder 2  <br>
2 ÷ 3 = 0 remainder 2 <br>
<br>

````python
def to_base_3(n):
    return format(n, 'b').replace('1', 'T') if n == 0 else ''

def decimal_to_base_3(n):
    digits = []
    while n > 0:
        digits.append(str(n % 3))
        n = n // 3
    return ''.join(reversed(digits))

# Testing
print("15 in base 3:", decimal_to_base_3(15))
print("24 in base 3:", decimal_to_base_3(24))
````

### Question 6.3.6
Write and test a function that counts and returns the number of bits (binary digits) set to 1 in a positive
integer n without converting n to a string.<br>
```python
def count_bits(n):
"""Return the count of bits set to 1 in a positive integer.
"""
```
For example, ```count_bits(12)``` should return 2, because 12 = 000011002.<br>
Hint: Check the digits from right to left; ```divmod(n, 2)``` returns n//2 and the rightmost binary digit;<br>
n//2 shifts the binary digits of n to the right by one and gets rid of the rightmost digit.
<br>
### Answer:
<br>

````python
def count_bits(n):
    """Return the count of bits set to 1 in a positive integer."""
    count = 0
    while n > 0:
        n, r = divmod(n, 2)
        if r == 1:
            count += 1
    return count

print(count_bits(12))  # should print 2
````

### Question 6.4.6
Find the fraction with a denominator less than or equal to 20 that best approximates π.
Hints:
1. Use the following to get the value of pi.
from math import pi
2. Recall that the built-in function abs() returns the absolute value of a number.
3. Use the following, or similar:
if approx < best_approx:<br>
best_approx = approx<br>
best_num, best_denom = num, denom
<br>

### Answer:
<br>


````python
from math import pi

best_approx = float('inf')
best_num = 0
best_denom = 1

for denom in range(1, 21):  # denominator from 1 to 20
    for num in range(1, denom * 4):  # range for numerator
        approx = abs(pi - num / denom)
        if approx < best_approx:
            best_approx = approx
            best_num = num
            best_denom = denom

print(f"Best approximation: {best_num}/{best_denom} ≈ {best_num / best_denom}")
````

### Question 6.5.1
Positive integers a, b, and c are said to form a Pythagorean triple if 𝑎2 + 𝑏2 = 𝑐2. It is fairly easy to show
that if we take any integers 0 < m < n , then 𝑎 = 𝑛2 − 𝑚2 , 𝑏 = 2mn , and form a Pythagorean triple.<br>
For example, if we take 𝑚 = 1 , n = 2 , we get 𝑎 = 3 , b = 4 . 𝑐 = 5.<br>
How many different pairs of integers 0 < m < n ≤ 7 are there?<br>
Write Python code that will generate and display Pythagorean triples for all such pairs of m and n.
<br>
### Answer:
<br>

````python
def pythagorean_triples(limit=7):
    triples = []
    for m in range(1, limit):
        for n in range(m+1, limit+1):
            a = n**2 - m**2
            b = 2*m*n
            c = n**2 + m**2
            triples.append((a, b, c))
    return triples

print(pythagorean_triples(7))
````
