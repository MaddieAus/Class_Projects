Hint: len(s) returns the number of characters in s; for example, len('La') returns 2. Do not use loops or other more advanced constructs. Keep it simple.<br>

````python
#Running code
def triangle(s):
    """
    Print an upside-down triangle using three print statements.
    Example: triangle('*')
    """
    print(s * 5)
    print(s * 3)
    print(s * 1)

triangle('*')
````

````python
def triangle(s):
    """
    Print an upside-down triangle using only one print statement.
    """
    print(s*5 + '\n' + s*3 + '\n' + s*1)

triangle('*')
````

````python
def triangle(s):
    """
    Print a symmetrical upside-down triangle even if s is a longer string.
    """
    print(s*5)
    print(s*3)
    print(s*1)

triangle("Ba")
````
