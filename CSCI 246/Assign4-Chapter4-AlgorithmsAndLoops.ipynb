### <u>Question 4.2.3:<u/>
Suppose you only have the operations of addition and subtraction. Devise an algorithm that takes two
positive integers, m and n, and calculates the quotient and the remainder when m is divided by n. (For
example, when 17 is divided by 5, the quotient is 3 and the remainder is 2.)
Present your algorithm in pseudocode.<br>

### <u>Anwser:</u>
```pseudocode
Algorithm Divide(m, n)
    quotient ← 0
    while m ≥ n do
        m ← m - n
        quotient ← quotient + 1
    end while
    remainder ← m
    return (quotient, remainder)
```

### <u>Question 4.3.2:</u>
Implement the algorithm from Example 1 in Section 4.2 (see below) in a Python function
```ApproxPIdiv4(n)```. Make your function take one argument, ```n```. Call the function for ```n = 1000,
10000, and 100000``` and compare each returned value to $$ \frac{\pi}{4} $$ .
```python
Input: n
k ← 1
sign ← 1
sum ← 0
while k < n:
sum ← sum + sign/k
k ← k + 2
sign ← -sign
Output: sum
```
<br>
Hint. Use the following to represent π:
from math import pi<br>

### <u>Anwser:</u>
```python
from math import pi

def ApproxPIdiv4(n):
    k = 1
    sign = 1
    total = 0
    while k < n:
        total += sign / k
        k += 2
        sign = -sign
    return total

# Test the function
for val in [1000, 10000, 100000]:
    result = ApproxPIdiv4(val)
    print(f"n = {val}, Approximation = {result}, π/4 = {pi/4}")
```

excluding the number itself. For example, 6 = 1 + 2 + 3. Write a function ```sum_of_divisors(n)```
that takes a positive integer n and returns the sum of all its divisors (excluding n).<br>
Hint:<br>

The smallest perfect number is 6. Use your ```function sum_of_divisors(n)```to find and print out the
next perfect number.<br>

### <u>Anwser:</u>
```python
def sum_of_divisors(n):
    total = 0
    for d in range(1, n):
        if n % d == 0:
            total += d
    return total

# Find the next perfect number
num = 7
while True:
    if sum_of_divisors(num) == num:
        print(f"The next perfect number is {num}")
        break
    num += 1
```
