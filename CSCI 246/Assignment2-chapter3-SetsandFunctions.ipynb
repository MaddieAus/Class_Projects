<b>Question 3.4.3</b>
Pretend that string’s rjust method does not exist and write your own:
```python
def right_justify(s, w):
"""Pad the given string s with spaces on the left to
form a new string of total length w and return
the new string.
"""
return ______
Hint: Recall that len(s) returns the length of string s.
Then test the following:
right_justify('123', 5)
right_justify(5, '123')
right_justify('123')
right_justify('12345', 3)<br>
```

<b>Results:</b><br>
```python
def right_justify(s, w):
    """Left-pad string s with spaces to total width w."""
    s = str(s) 
    if w <= len(s):
        return s
    return ' ' * (w - len(s)) + s
```

(a) The quadratic equation 𝑎𝑥2 + 𝑏𝑥 + 𝑐 = 0 may have two roots:
𝑥1 = −𝑏+√𝑏2−4𝑎𝑐 2𝑎 and 𝑥2 = −𝑏−√𝑏2−4𝑎𝑐 2𝑎 .<br>
Write and test a function that takes a, b, and c as arguments and returns two values that are the values
of the two roots.<br>
• Import sqrt:<br>
from math import sqrt<br>
• Calculate √𝑏2 − 4𝑎𝑐 only once.<br>
• Leave it to sqrt to raise an exception when 𝑏2 − 4𝑎𝑐 < 0 . (You do not need to handle this
exception.)<br>

<b>Results:</b><br>
```python
from math import sqrt

def quad_roots(a, b, c):
    d = b*b - 4*a*c
    r = sqrt(d)
    return (-b + r)/(2*a), (-b - r)/(2*a)
```

(b) Create a new version of your function from Part (a) that returns None if 𝑏2 − 4𝑎𝑐 < 0 .<br>

<b>Results:</b><br>
```python
def quad_roots_or_none(a, b, c):
    d = b*b - 4*a*c
    if d < 0:
        return None
    r = sqrt(d)
    return (-b + r)/(2*a), (-b - r)/(2*a)
```

(c) Prompt the user to enter three real numbers, a, b, and c, and then call the function from Part (b), and
display the result. If the function returns None, display “No real number solutions.”<br>

<b>Results:</b><br>
```python
a = float(input("a: "))
b = float(input("b: "))
c = float(input("c: "))
ans = quad_roots_or_none(a, b, c)
print(ans if ans is not None else "No real number solutions.")
```
