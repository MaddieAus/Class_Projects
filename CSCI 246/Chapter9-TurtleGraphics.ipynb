````python
import ipyturtle3 as turtle
import math, random
from IPython.display import display

def make_turtle_canvas():
    """Creates and returns a new turtle + screen on a fresh canvas."""
    canvas = turtle.Canvas()
    display(canvas)
    screen = turtle.TurtleScreen(canvas)
    t = turtle.Turtle(screen)
    return t
````

### Question 9.2.2
Write a function draw_polygon(t, n, a) that uses a turtle t to draw a regular polygon with n sides
of length a, in the current color. <br>
Hint: if you have n sides, you need to turn n times and cover the whole 360° angle at the end, so each turn is 360 𝑛 degrees. <br>
Demonstrate that your function works by calling it to draw several polygons within a single Canvas.
### Answer:

````python
def draw_polygon(t, n, a):
    """Draw a regular polygon with n sides of length a using turtle t."""
    t.pendown()
    for i in range(n):
        t.forward(a)
        t.left(360 / n)
    t.penup()

# Demonstration
t = make_turtle_canvas()
colors = ["red", "blue", "green", "purple", "orange"]
for i, sides in enumerate([3, 4, 5, 6, 8]):
    t.color(colors[i])
    t.goto(-200 + i*100, 0)
    draw_polygon(t, sides, 50)
````

````python
def draw_rectangle(t, w, h):
    """Draw a rectangle of width w and height h using turtle t."""
    t.pendown()
    for k in range(2):
        t.forward(w)
        t.left(90)
        t.forward(h)
        t.left(90)
    t.penup()

def draw_flagstone_path(t, n=10):
    """Draws n flagstones that bend slightly upward."""
    colors = ["red", "blue", "darkgreen", "black", "orange"]
    t.penup()
    t.goto(-200, -100)
    for i in range(n):
        t.color(random.choice(colors))
        draw_rectangle(t, 40, 20)
        t.forward(40)
        t.left(5)   # bend slightly upward

t = make_turtle_canvas()
draw_flagstone_path(t)
````

### Answer:

````python
def draw_triangle(t, a):
    """Draw an equilateral triangle with side length a."""
    t.pendown()
    for k in range(3):
        t.forward(a)
        t.left(120)
    t.penup()

def draw_triangle_pyramid(t, levels=4, size=60):
    """Draws a pyramid of equilateral triangles."""
    t.penup()
    start_x, start_y = -150, -100
    for row in range(levels):
        t.goto(start_x + row * (size / 2), start_y + row * (size * math.sin(math.radians(60))))
        for col in range(levels - row):
            draw_triangle(t, size)
            t.forward(size)
        t.penup()

t = make_turtle_canvas()
draw_triangle_pyramid(t)
````

### Question 9.2.7
Write and test a function to draw a flower with a specified number of petals:
Hint: For a prettier flower, the number of petals should be an odd number. Add 1 if the specified number
of petals is even.
### Answer:

````python
def draw_flower(t, petals=7, radius=50):
    """Draws a flower with a specified number of petals."""
    if petals % 2 == 0:
        petals += 1  # make odd for symmetry
    t.color("magenta")
    for _ in range(petals):
        t.circle(radius, 60)
        t.left(120)
        t.circle(radius, 60)
        t.left(360 / petals)

t = make_turtle_canvas()
draw_flower(t, petals=9)
````

### Question 9.3.2
The game of Nim is, theoretically, played with piles of stones, but it is commonly played with rows of sticks
instead. Draw a configuration with three rows of sticks:
### Answer:

````python
def draw_stick(t, length=30):
    t.pendown()
    t.forward(length)
    t.backward(length)
    t.penup()

def draw_nim_sticks(t):
    """Draw 3 rows of sticks like the game of Nim."""
    t.penup()
    start_x, start_y = -100, 0
    for row, num_sticks in enumerate([3, 5, 7]):
        t.goto(start_x, start_y - row * 40)
        for i in range(num_sticks):
            draw_stick(t)
            t.forward(20)

t = make_turtle_canvas()
draw_nim_sticks(t)
````

### Question 9.3.4
Draw a snowman:
### Answer:

````python
def draw_snowman(t):
    """Draw a simple snowman with three stacked circles."""
    t.penup()
    t.goto(0, -100)
    for r in [40, 30, 20]:
        t.pendown()
        t.circle(r)
        t.penup()
        t.left(90)
        t.forward(r*2)
        t.right(90)

t = make_turtle_canvas()
draw_snowman(t)
````

### Question 9.3.8
Draw a fairly smooth graph of the parabola 𝑦 = 𝑥2 along with the x- and y-axis:
Hint: generate the segment of the parabola for −3 ≤ 𝑥 ≤ 3 but scale the graph by a factor of 30 or 40.
### Answer:

````python
def draw_parabola(t):
    """Draw y = x^2 scaled by 30, for -3 <= x <= 3."""
    t.penup()
    scale = 30
    points = [(x, (x**2)) for x in [i/10 for i in range(-30, 31)]]
    
    # Draw axes
    t.goto(-100, 0)
    t.pendown()
    t.forward(200)
    t.penup()
    t.goto(0, -50)
    t.left(90)
    t.pendown()
    t.forward(100)
    t.penup()
    t.right(90)
    
    # Plot parabola
    for x, y in points:
        t.goto(x*scale, y*scale/4 - 50)
        t.pendown()
    t.penup()

t = make_turtle_canvas()
draw_parabola(t)
````

````python
def draw_color_swatches(t):
    values = [0, 127, 255]
    t.penup()
    y = 0
    for r in values:
        for g in values:
            for b in values:
                t.goto((b/10) - 150, y - 100)
                t.color((r, g, b))
                t.begin_fill()
                for _ in range(4):
                    t.forward(20)
                    t.left(90)
                t.end_fill()
            y += 25
        y += 30

t = make_turtle_canvas()
draw_color_swatches(t)
````
